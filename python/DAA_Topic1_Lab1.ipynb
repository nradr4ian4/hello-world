```python
urls = ["site.com/page1", "site.com/page2", "BROKEN", "site.com/page4", "BLOCKED", "site.com/page6"]

for url in urls:
  if url == "BROKEN":
    print("Skipping a broken link")
    continue
  elif url =="BLOCKED":
    print("Server blocked us, stopping the scraper")
    break
  else:
    print("Scraping",url)
```

```python
# while loop: keep doubling savings until they pass a target
savings, months = 100,0
while savings <1000:
  savings = savings*2
  months=months+1
print(f"Reached {savings} after {months} months")
```

```python
raw_csv = """product,units,price
Notebook,12,3.50
Pen,40,1.20
Backpack,3,45.00
Calculator,5,22.90"""

rows=raw_csv.split("\n")
header=rows[0].split(",")
print("Columns: ",header)

total_revenue = 0
prices = []

for row in rows[1:]: #start at index 1 aka 2nd row
  product, units, price = row.split(",")
  units=int(units)
  price=float(price)
  revenue=units*price
  prices.append(price)
  total_revenue += revenue
  print(f"{product:<11} {units:>3} x {price:>6.2f} = {revenue:>7.2f}")

print(f"Total revenue: {total_revenue:.2f}")                      # 4. statistics
print(f"Average price: {sum(prices) / len(prices):.2f}")
```

```python
students = [
    {"name": "Aisha", "score": 88},
    {"name": "Ben", "score": 64},
    {"name": "Chen", "score": 91},
    {"name": "Divya", "score": 70},
    {"name": "Ethan", "score": 55},
]

score=[s["score"] for s in students]
average = sum(score) / len(scores)

passed =0
for s in students:
  if s["score"] >=70:
    passed+=1

print(f"Students:  {len(students)}")
print(f"Average score: {average:.1f}")
print(f"Passed: {passed} | Failed: {len(students) - passed}")
print("Top scorer: ", max(students, key=lambda s: s["score"])["name"])
#print("Top scorer:", max(students, key=lambda s: s["score"])["name"])
```
